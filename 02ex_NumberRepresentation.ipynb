1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
def converter(x_int):
    x = abs(x_int)

    # Decimal to Binary
    bin_list = []
    q2, r2 = divmod(x, 2)
    bin_list.append(r2)
    while q2 != 0:
        q2, r2 = divmod(q2, 2)
        bin_list.append(r2)
    while len(bin_list) < 8:
        bin_list.append(0)

    binary = bin_list[::-1]
    binary_str = "".join(map(str, bin_list[::-1]))
    if x_int < 0:
        binary[0] = 1
        print('Binary representation of', x, ':', "-" + binary_str)
    else:
        print('Binary representation of', x, ':', binary_str)


    # Decimal to Hexadecimal
    hex_list = []
    q16, r16 = divmod(x, 16)
    hex_list.append(r16)
    while q16 != 0:
        q16, r16 = divmod(q16, 16)
        hex_list.append(r16)

    hexadecimal = "".join(map(str, hex_list[::-1]))
    if x_int < 0:
        print('Hexadecimal representation of', x, ':', "-0x" + hexadecimal)
    else:
        print('Hexadecimal representation of', x, ':', "0x" + hexadecimal)


    # Binary to Decimal
    dec = 0
    for i, value in enumerate(binary[::-1]):
        dec += value * 2**i
        if i == 7 and value == 1:
            dec = (-1) * (dec - (value * 2**7))
    print('Decimal representation of', binary_str, ':', dec)


    # Hexadecimal to Decimal
    dec2 = 0
    for i, value in enumerate(hex_list):
        dec2 += value * 16**i
    if x_int < 0:
        print('Decimal representation of', "0x" + hexadecimal, ':', "-" + str(dec2))
    else:
        print('Decimal representation of', "0x" + hexadecimal, ':', dec2)
```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
def bit_convert(x_str):
    sign_bit = int(x_str[0])

    exp_bit = int(x_str[1:9], 2)
    exp_bias = 2**(exp_bit - 127)

    mantissa_bits = x_str[9:]
    mantissa = 1
    for i, x in enumerate(mantissa_bits):
        mantissa += int(x) * 2**(-(i+1))

    print((-1)**(sign_bit) * mantissa * exp_bias)
```

```python
# input should be a 32 bit string without spaces and other characters
x_str = '11000000101100000000000000000000'
bit_convert(x_str)
```

3\. Write a program to determine the underflow and overflow limits (within a factor of 2) for python on your computer. 

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
# underflow
a = 1.0
try:
    while a > 0.0:
        a_prev = a
        a = a/2.0
    print("Last value before underflow: ", a_prev)
except:
    print("Underflow reached")
```

```python
# overflow
a = 1.0
try:
    while a+1 != a:
        a_prev = a
        a = a*2.0
    print("Last value before overflow: ", a_prev)
except:
    print("Overflow reached")
```

6\. Write a program that implements the function $f(x)=x(x−1)$

(a) Calculate the derivative of the function at the point $x = 1$ using the derivative definition:

$$
\frac{{\rm d}f}{{\rm d}x} = \lim_{\delta\to0} \frac{f(x+\delta)-f(x)}{\delta}
$$

with $\delta = 10^{−2}$. Calculate the true value of the same derivative analytically and compare with the answer your program gives. The two will not agree perfectly. Why not?

(b) Repeat the calculation for $\delta = 10^{−4}, 10^{−6}, 10^{−8}, 10^{−10}, 10^{−12}$ and $10^{−14}$. How does the accuracy scales with $\delta$?

```python
fxn = lambda x: x*(x-1)

# (a)
x = 1
delta_a = 10**(-2)
derivative_a = (fxn(x + delta_a) - fxn(x)) / delta_a
print("My program derivative: ", derivative_a)
```

**Analytical solution:**

$$
f(x) = x(x-1) \\
f'(x) = (x-1) + x
$$

for $x=1$
$$ f'(1) = (1-1) + 1 \\
\boxed{f'(1) = 1}
$$

Analytically, the derivative of the given function for $x=1$ is equal to $1$. But numerically, when the derivative is calculated using its limit definition with $\delta = 10^{-2}$, the program results to $1.010000000000001$. The two values do not agree perfectly because of rounding errors in computer arithmetic which is due to the fact that the machine can only store numbers with finite precision.

```python
# (b)
derivative_true = 1.000000000000000
delta_b = [10**(-4), 10**(-6), 10**(-8), 10**(-10), 10**(-12), 10**(-14)]
for delta in delta_b:
    derivative_b = 0
    accuracy_der = 0

    derivative_b = (fxn(x + delta) - fxn(x)) / delta
    accuracy_der = 100 - (abs(derivative_b - derivative_true)/derivative_true * 100)
    print("Sigma: ", delta, "\nDerivative: ", derivative_b, "\nAccuracy: ", accuracy_der, "%\n")
```

**Question:** How does the accuracy scales with $\delta$? \
**Answer:** Initially, as the sigma is increased, the accuracy also increases. However, when sigma is $10^{-10}$ or smaller, the accuracy starts to decrease.

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k$-th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute? 


```python
import numpy as np
import math
import time

# (a)
N = 100
fxn2 = lambda x: (1-(x**2))**(1/2)
points = np.linspace(-1, 1, N)
h = 2/N

start_time = time.time()
integral_a = h * sum(map(fxn2, points))
end_time = time.time()
print("--- %s seconds ---" % (end_time - start_time))
print("My program integral: ", integral_a)

integral_true = math.pi / 2
accuracy_int = 100 - (abs(integral_a - integral_true)/integral_true * 100)
print("True Value: ", integral_true)
print("Accuracy: ", accuracy_int, "%")
```

```python
# (b)
runtime = 0
integral_b = 0
N = 150000
while runtime < 0.1:
    points = np.linspace(-1, 1, N)
    integral_prev = integral_b
    runtime_prev = runtime
    N_prev = N

    start_time = time.time()
    integral_b = (2/N) * sum(map(fxn2, points))
    end_time = time.time()

    runtime = end_time - start_time
    N += 1
    if N_prev % 100 == 0:
        print("check: ", N_prev)

print("--- %s seconds ---" % runtime_prev)
print("Maximum N: ", N_prev)
print("Program integral value: ", integral_prev)
```
